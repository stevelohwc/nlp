## 1. Text Preprocessing

Text preprocessing is crucial for NLP tasks. We'll demonstrate various preprocessing techniques.

```python
# Initialize preprocessor
preprocessor = TextPreprocessor()

# Sample text
sample_text = """Hello! Check out https://example.com for more information. 
The Natural Language Processing course is AMAZING!!! Contact us at info@example.com"""

print("Original text:")
print(sample_text)
print("\n" + "="*70 + "\n")

# Clean text
cleaned = preprocessor.clean_text(sample_text)
print("Cleaned text:")
print(cleaned)
print("\n" + "="*70 + "\n")

# Tokenize
tokens = preprocessor.tokenize(cleaned)
print("Tokens:")
print(tokens)
print("\n" + "="*70 + "\n")

# Remove stopwords
filtered_tokens = preprocessor.remove_stopwords(tokens)
print("Without stopwords:")
print(filtered_tokens)
print("\n" + "="*70 + "\n")

# Complete preprocessing
preprocessed = preprocessor.preprocess(sample_text)
print("Fully preprocessed:")
print(preprocessed)
```

## 2. Sentiment Analysis

Analyze the sentiment of text documents.

```python
# Initialize sentiment analyzer
analyzer = SentimentAnalyzer()

# Sample texts
texts = [
    "I love this project! It's absolutely fantastic!",
    "This is terrible. I'm very disappointed.",
    "It's okay, nothing special.",
    "Amazing work! Highly recommended!",
    "Not good at all. Waste of time."
]

print("Sentiment Analysis Results:")
print("="*70)

for i, text in enumerate(texts, 1):
    result = analyzer.analyze(text)
    print(f"\nText {i}: {text}")
    print(f"Sentiment: {result['label'].upper()}")
    print(f"Compound Score: {result['compound']:.3f}")
    print(f"Positive: {result['positive']:.3f} | Negative: {result['negative']:.3f} | Neutral: {result['neutral']:.3f}")
```

```python
# Get sentiment distribution
distribution = analyzer.get_sentiment_distribution(texts)

print("\nSentiment Distribution:")
print("="*70)
for sentiment, count in distribution.items():
    print(f"{sentiment.capitalize()}: {count}")
```

## 3. Text Classification

Train a classifier to categorize text documents.

```python
# Training data - movie reviews
train_texts = [
    "This movie is fantastic! Great acting and amazing plot.",
    "Terrible film. Waste of time and money.",
    "I loved every minute of this movie. Highly recommended!",
    "Boring and predictable. Not worth watching.",
    "Excellent cinematography and brilliant performances.",
    "Poor script and bad direction. Very disappointed.",
    "One of the best movies I've seen this year!",
    "Awful movie. I walked out halfway through.",
    "Masterpiece! The director is a genius.",
    "Complete disaster. Nothing worked in this film.",
    "Beautiful story with great character development.",
    "Waste of talent. Even good actors couldn't save it.",
    "Incredible movie experience. Must watch!",
    "Disappointing and frustrating to watch.",
    "Amazing from start to finish. Loved it!",
    "Terrible acting and boring storyline.",
]

train_labels = [
    'positive', 'negative', 'positive', 'negative',
    'positive', 'negative', 'positive', 'negative',
    'positive', 'negative', 'positive', 'negative',
    'positive', 'negative', 'positive', 'negative'
]

# Create and train classifier
classifier = TextClassifier(classifier_type='naive_bayes', vectorizer_type='tfidf')
results = classifier.train(train_texts, train_labels, test_size=0.25)

print("Training Results:")
print("="*70)
print(f"Accuracy: {results['accuracy']:.3f}")
print(f"\nClassification Report:")
print(results['classification_report'])
```

```python
# Test with new examples
test_texts = [
    "This is a wonderful film with great performances!",
    "I didn't enjoy this movie at all. Very boring.",
    "Absolutely brilliant! One of my favorites."
]

predictions = classifier.predict(test_texts)
probabilities = classifier.predict_proba(test_texts)

print("\nPredictions on New Data:")
print("="*70)

for i, (text, pred, proba) in enumerate(zip(test_texts, predictions, probabilities), 1):
    print(f"\nTest {i}:")
    print(f"Text: {text}")
    print(f"Prediction: {pred.upper()}")
    print(f"Confidence: {max(proba):.3f}")
```
